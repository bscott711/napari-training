```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# Parameters
grid_size = (100, 100)  # Size of the 2D grid
bleach_radius = 10      # Radius of the bleached ROI
diffusion_rate = 0.1    # Diffusion rate (arbitrary units)
time_steps = 50         # Number of time steps to simulate
noise_level = 0.01      # Add noise to simulate real-world conditions

# Initialize the grid with uniform fluorescence
grid = np.ones(grid_size)

# Create a circular bleached ROI
center = (grid_size[0] // 2, grid_size[1] // 2)
y, x = np.ogrid[:grid_size[0], :grid_size[1]]
mask = ((x - center[1])**2 + (y - center[0])**2) <= bleach_radius**2
grid[mask] = 0.1  # Bleach the ROI (reduce fluorescence)

# Simulate diffusion over time
frames = [grid.copy()]
for t in range(1, time_steps):
    # Apply Gaussian smoothing to simulate diffusion
    grid = gaussian_filter(grid, sigma=diffusion_rate)
    
    # Add noise to simulate experimental conditions
    grid += np.random.normal(0, noise_level, grid_size)
    grid = np.clip(grid, 0, 1)  # Ensure values stay within [0, 1]
    
    frames.append(grid.copy())

# Convert frames to a 3D numpy array (time, height, width)
frames = np.array(frames)

# Save the simulated FRAP data as a TIFF stack
from tifffile import imsave
imsave('frap_simulation.tif', frames)

print("FRAP simulation saved as 'frap_simulation.tif'")
```
